````python
import os, json, time
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix
from groq import Groq
from google.colab import userdata

client = Groq(api_key=userdata.get("GROQ_API_KEY"))

CATEGORIES = [
    "Billing & Payments",
    "Technical Issue",
    "Account Access",
    "Shipping & Delivery",
    "Product Defect",
    "Feature Request",
    "Refund & Returns",
]
````

````python

raw_tickets = [
    # Billing
    ("I was charged twice for my subscription this month.", "Billing & Payments"),
    ("My invoice shows the wrong amount.", "Billing & Payments"),
    ("I can't update my credit card details.", "Billing & Payments"),
    ("Why was I billed after I cancelled?", "Billing & Payments"),
    ("I need a VAT receipt for my last payment.", "Billing & Payments"),
    ("Promo code didn't apply to my order.", "Billing & Payments"),
    ("Charged for a plan I didn't upgrade to.", "Billing & Payments"),
    # Technical
    ("The app crashes every time I open it on iPhone.", "Technical Issue"),
    ("Login page keeps loading forever.", "Technical Issue"),
    ("Error 500 when I try to export my data.", "Technical Issue"),
    ("Push notifications stopped working.", "Technical Issue"),
    ("Dark mode is broken on Android 14.", "Technical Issue"),
    ("Can't upload files larger than 10 MB.", "Technical Issue"),
    ("Search results are not loading.", "Technical Issue"),
    # Account Access
    ("I forgot my password and the reset email never arrived.", "Account Access"),
    ("My account was locked after too many login attempts.", "Account Access"),
    ("Two-factor authentication code is not working.", "Account Access"),
    ("I can't change my email address in settings.", "Account Access"),
    ("My account was hacked, please help.", "Account Access"),
    ("SSO login with Google is broken.", "Account Access"),
    ("Can't delete my account.", "Account Access"),
    # Shipping
    ("My order hasn't arrived after 2 weeks.", "Shipping & Delivery"),
    ("Tracking number shows delivered but I got nothing.", "Shipping & Delivery"),
    ("Wrong item was delivered.", "Shipping & Delivery"),
    ("Package arrived damaged.", "Shipping & Delivery"),
    ("Can I change my delivery address?", "Shipping & Delivery"),
    ("Express shipping option wasn't available at checkout.", "Shipping & Delivery"),
    ("Order stuck in customs.", "Shipping & Delivery"),
    # Product Defect
    ("The screen on my device has dead pixels.", "Product Defect"),
    ("Battery drains completely within 2 hours.", "Product Defect"),
    ("Buttons stopped responding after first week.", "Product Defect"),
    ("Device overheats during normal use.", "Product Defect"),
    ("Speaker produces crackling sound.", "Product Defect"),
    ("Zipper broke on the bag I just received.", "Product Defect"),
    ("Paint chipped off after a single wash.", "Product Defect"),
    # Feature Request
    ("Please add dark mode to the web app.", "Feature Request"),
    ("Would love an export to CSV feature.", "Feature Request"),
    ("Can you add support for multiple currencies?", "Feature Request"),
    ("Please integrate with Slack.", "Feature Request"),
    ("Bulk delete option for messages would be great.", "Feature Request"),
    ("Add keyboard shortcuts for power users.", "Feature Request"),
    ("I'd like to schedule reports to be sent by email.", "Feature Request"),
    # Refund
    ("I'd like a refund for my unused subscription.", "Refund & Returns"),
    ("Product doesn't match the description, want a return.", "Refund & Returns"),
    ("Refund was promised 10 days ago and hasn't appeared.", "Refund & Returns"),
    ("How do I return a gift?", "Refund & Returns"),
    ("Received wrong size, need exchange or refund.", "Refund & Returns"),
    ("I changed my mind, can I get my money back?", "Refund & Returns"),
    ("Partial refund applied — expected full amount.", "Refund & Returns"),
]

df = pd.DataFrame(raw_tickets, columns=["ticket", "true_label"])
print(f"Dataset: {len(df)} tickets across {df['true_label'].nunique()} categories")
df["true_label"].value_counts().plot(kind="bar", figsize=(9,4), title="Ticket Distribution")
plt.tight_layout(); plt.show()
````

````python
SYSTEM_ZERO_SHOT = """You are a support ticket classification expert.
Given a support ticket, return the top 3 most probable categories in ranked order.

Available categories:
- Billing & Payments
- Technical Issue
- Account Access
- Shipping & Delivery
- Product Defect
- Feature Request
- Refund & Returns

Respond ONLY with a JSON object like:
{"top3": ["Category1", "Category2", "Category3"]}
No explanation. No extra text. No markdown."""

def classify_zero_shot(ticket_text):
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",   # Using an available model
        messages=[
            {"role": "system", "content": SYSTEM_ZERO_SHOT},
            {"role": "user",   "content": ticket_text}
        ],
        temperature=0,
        max_tokens=100,
    )
    raw = response.choices[0].message.content.strip()
    # clean any accidental markdown fences
    raw = raw.replace("```json"," ").replace("```"," ").strip()
    return json.loads(raw)["top3"]

# quick test
print(classify_zero_shot(df["ticket"].iloc[0]))
````

````python
FEW_SHOT_EXAMPLES = """
Example 1:
Ticket: "My credit card was charged but I never received a confirmation email."
Response: {"top3": ["Billing & Payments", "Technical Issue", "Account Access"]}

Example 2:
Ticket: "The app freezes on the dashboard every morning."
Response: {"top3": ["Technical Issue", "Account Access", "Feature Request"]}

Example 3:
Ticket: "I want to return the jacket I ordered, it's the wrong colour."
Response: {"top3": ["Refund & Returns", "Shipping & Delivery", "Product Defect"]}

Example 4:
Ticket: "Can't log in, reset link expired instantly."
Response: {"top3": ["Account Access", "Technical Issue", "Billing & Payments"]}

Example 5:
Ticket: "Would be amazing if you added Zapier integration."
Response: {"top3": ["Feature Request", "Technical Issue", "Account Access"]}
"""

SYSTEM_FEW_SHOT = f"""You are a support ticket classification expert.
Given a support ticket, return the top 3 most probable categories in ranked order.

Available categories:
- Billing & Payments
- Technical Issue
- Account Access
- Shipping & Delivery
- Product Defect
- Feature Request
- Refund & Returns

Here are some examples:
{FEW_SHOT_EXAMPLES}

Respond ONLY with a JSON object like:
{{"top3": ["Category1", "Category2", "Category3"]}}
No explanation. No extra text. No markdown."""

def classify_few_shot(ticket_text):
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",  # Using available model
        messages=[
            {"role": "system", "content": SYSTEM_FEW_SHOT},
            {"role": "user",   "content": ticket_text}
        ],
        temperature=0,
        max_tokens=100,
    )
    raw = response.choices[0].message.content.strip()
    raw = raw.replace("```json"," ").replace("```"," ").strip()
    return json.loads(raw)["top3"]

print(classify_few_shot(df["ticket"].iloc[0]))
````

````python
FEW_SHOT_EXAMPLES = """
Example 1:
Ticket: "My credit card was charged but I never received a confirmation email."
Response: {"top3": ["Billing & Payments", "Technical Issue", "Account Access"]}

Example 2:
Ticket: "The app freezes on the dashboard every morning."
Response: {"top3": ["Technical Issue", "Account Access", "Feature Request"]}

Example 3:
Ticket: "I want to return the jacket I ordered, it's the wrong colour."
Response: {"top3": ["Refund & Returns", "Shipping & Delivery", "Product Defect"]}

Example 4:
Ticket: "Can't log in, reset link expired instantly."
Response: {"top3": ["Account Access", "Technical Issue", "Billing & Payments"]}

Example 5:
Ticket: "Would be amazing if you added Zapier integration."
Response: {"top3": ["Feature Request", "Technical Issue", "Account Access"]}
"""

SYSTEM_FEW_SHOT = f"""You are a support ticket classification expert.
Given a support ticket, return the top 3 most probable categories in ranked order.

Available categories:
- Billing & Payments
- Technical Issue
- Account Access
- Shipping & Delivery
- Product Defect
- Feature Request
- Refund & Returns

Here are some examples:
{FEW_SHOT_EXAMPLES}

Respond ONLY with a JSON object like:
{{"top3": ["Category1", "Category2", "Category3"]}}
No explanation. No extra text. No markdown."""

def classify_few_shot(ticket_text):
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {"role": "system", "content": SYSTEM_FEW_SHOT},
            {"role": "user",   "content": ticket_text}
        ],
        temperature=0,
        max_tokens=100,
    )
    raw = response.choices[0].message.content.strip()
    raw = raw.replace("```json","").replace("```","").strip()
    return json.loads(raw)["top3"]

print(classify_few_shot(df["ticket"].iloc[0]))
````

````python
zero_shot_results, few_shot_results = [], []

for i, row in df.iterrows():
    try:
        zs = classify_zero_shot(row["ticket"])
        fs = classify_few_shot(row["ticket"])
    except Exception as e:
        print(f"Row {i} error: {e}")
        zs = fs = ["Unknown", "Unknown", "Unknown"]

    zero_shot_results.append(zs)
    few_shot_results.append(fs)
    time.sleep(0.3)   # respect free tier rate limits

    if i % 10 == 0:
        print(f"Progress: {i+1}/{len(df)}")

df["zero_shot_top3"] = zero_shot_results
df["few_shot_top3"]  = few_shot_results
df["zero_shot_top1"] = [r[0] for r in zero_shot_results]
df["few_shot_top1"]  = [r[0] for r in few_shot_results]

print("\nSample predictions:")
print(df[["ticket","true_label","zero_shot_top1","few_shot_top1"]].head(10).to_string())
````

````python
zs_top1 = (df["zero_shot_top1"] == df["true_label"]).mean()
fs_top1 = (df["few_shot_top1"]  == df["true_label"]).mean()

zs_top3 = df.apply(lambda r: r["true_label"] in r["zero_shot_top3"], axis=1).mean()
fs_top3 = df.apply(lambda r: r["true_label"] in r["few_shot_top3"],  axis=1).mean()

print(f"Zero-Shot  →  Top-1: {zs_top1:.2%}   Top-3: {zs_top3:.2%}")
print(f"Few-Shot   →  Top-1: {fs_top1:.2%}   Top-3: {fs_top3:.2%}")
````

````python
print("="*40)
print("ZERO-SHOT")
print(classification_report(df["true_label"], df["zero_shot_top1"], zero_division=0))

print("="*40)
print("FEW-SHOT")
print(classification_report(df["true_label"], df["few_shot_top1"], zero_division=0))
````

````python
fig, axes = plt.subplots(1, 2, figsize=(18, 7))

for ax, col, title in zip(axes,
    ["zero_shot_top1", "few_shot_top1"],
    ["Zero-Shot (LLaMA3 on Groq)", "Few-Shot (LLaMA3 on Groq)"]):

    cm = confusion_matrix(df["true_label"], df[col], labels=CATEGORIES)
    sns.heatmap(cm, annot=True, fmt="d",
                xticklabels=CATEGORIES, yticklabels=CATEGORIES,
                cmap="YlOrRd", ax=ax)
    ax.set_title(title, fontsize=12)
    ax.set_xlabel("Predicted"); ax.set_ylabel("True")
    ax.tick_params(axis="x", rotation=45)

plt.tight_layout(); plt.show()
````

````python
pd.DataFrame({
    "Method":    ["Zero-Shot", "Few-Shot"],
    "Top-1 Acc": [zs_top1, fs_top1],
    "Top-3 Acc": [zs_top3, fs_top3],
}).set_index("Method").plot(
    kind="bar", figsize=(8,5),
    color=["#2196F3","#FF5722"],
    edgecolor="black"
)
plt.title("Groq LLaMA3 — Zero-Shot vs Few-Shot")
plt.ylim(0, 1.05); plt.ylabel("Accuracy"); plt.xticks(rotation=0)
plt.tight_layout(); plt.show()
````
